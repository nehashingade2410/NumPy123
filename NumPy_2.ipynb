```python
import numpy as np
```

```python
# Q. what is difference between reshape and resize?

# reshape() --changes the shapes of an array
# resize() -- changes the size and shape of an array

# reshape() --does not change the original array
# resize() --can change the original array

# reshape() --Total number of elements must remain the same
# resize() --Number of elements can increase or decrease

# reshape() -- Returns a new reshaped array
# resize() -- Can modify the original array when used as a method
```

```python
# resize does changes permanently
# reshape can be used for temporarly
```

```python
a = np.arange(24)
a
```

```python
a.resize((4,6))
a
```

```python
a.reshape(24)
```

```python
# 1. add a number in 1D array
# 2. add a row in 2D array
# 3. add a column in 2D array
# 4. remove a number from 1D array
# 5. remove a row from 2D array
# 6. remove a column from 2D array
# 7. Concatenate 2 arrays
```

```python
# 1. add a number in 1D array
a = np.array([1,2,3,4])
a = np.append(a,5)
a
```

```python
# 2. add a row in 2D array

a = np.array([[1, 2, 3, 4],
              [5, 6, 7, 8]])
new_row = [9, 0, 2, 3]

a = np.append(a,[new_row],axis =0)
a
```

```python
# axis=0 → add a row
# axis=1 → add a column
```

```python
# 3. add a column in 2D array

a = np.array([[1, 2, 3],
              [4, 5, 6]])

new_column = [7, 8]

a = np.append(a, np.array(new_column).reshape(2,1), axis =1)

print(a)
```

```python
# 4. remove a number from 1D array

a = np.array([1,2,3,4])
a = np.delete(a,2)  # here 2 is the index, so 3 will be removed
a
```

```python
# 5. remove a row from 2D array

a = np.array([[1, 2, 3],
              [4, 5, 6],
              [5, 6, 2]])
b = np.delete(a,2,axis=0)  # 2 is index,so [5, 6, 2] will be deleted
b
```

```python
# 6. remove a column from 2D array

a = np.array([[1, 2, 3],
              [4, 5, 6],
              [5, 6, 2]])
b = np.delete(a,1,axis=1)  # column at index 1,so [2,
                                              #    5,
                                              #    6,] will be deleted
b
```

```python
# 7. Concatenate 2 arrays
# np.concatenate() to join two arrays

a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([[7, 8, 9],
              [10, 11, 12]])

c = np.concatenate((a,b))
print(c)
```

```python
a = np.array([1,3,5])
b = np.array([2,4,6])
x = np.concatenate((a,b))
x
```

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([[7, 8, 9],
              [10, 11, 12]])

c = np.concatenate((a,b),axis =1)
print(c)
```

```python
# axis = 0 --concatenate rows(vertically)
# axis = 1 --concatenate columns(horizontally)
```
