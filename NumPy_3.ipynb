```python
import numpy as np
```

```python
# Identity matrix - diagonal elements are 1 and non diagonal
# elements are 0
# this will create a scalar matrix of same rows and columns
print(np.eye(4,dtype =int))
```

```python
print(np.eye(4,dtype=bool))
```

```python
print(np.eye(4,M=3,k=1,dtype=int))
```

```python
# there is another method in numpy to create an identity matrix

np.identity(3)
```
